```python
from pathlib import Path
import json, time

hits = sorted(Path('/kaggle/input').rglob('s114_source.ipynb'))
if not hits:
    raise FileNotFoundError('s114_source.ipynb not found in /kaggle/input')
source_path = hits[0]
print('Executing source notebook:', source_path)
nb = json.load(open(source_path))
start = time.time()
for i, cell in enumerate(nb.get('cells', [])):
    if cell.get('cell_type') != 'code':
        continue
    src = cell.get('source', '')
    if isinstance(src, list):
        src = ''.join(src)
    if not src.strip():
        continue
    print(f'--- run source cell {i} ---')
    result = get_ipython().run_cell(src)
    if result.error_before_exec is not None:
        raise result.error_before_exec
    if result.error_in_exec is not None:
        raise result.error_in_exec
print(f'S114 launcher complete in {(time.time()-start)/60:.2f} min')
```
